## Задание 14

```python
above_25 = df[df["T"] > 25]
min_co = above_25["CO(GT)"].min()
print("Минимальный уровень CO(GT):", min_co)
above_25[above_25["CO(GT)"] == min_co]
```

## Задание 15

```python
high_humidity = df[df["RH"] > 90]
print("Количество измерений:", len(high_humidity))
high_humidity
```

## Задание 16

```python
mean_co_above = df.loc[df["T"] > 20, "CO(GT)"].mean()
mean_co_below = df.loc[df["T"] < 20, "CO(GT)"].mean()
difference = round(mean_co_above - mean_co_below, 2)

print("Средний CO(GT) при T > 20:", round(mean_co_above, 2))
print("Средний CO(GT) при T < 20:", round(mean_co_below, 2))
print("Разница:", difference)
```

## Задание 17

```python
mean_ozone = df["PT08.S5(O3)"].mean()
df["High_Ozone"] = (df["PT08.S5(O3)"] > mean_ozone).astype(int)
print("Средний уровень озона:", round(mean_ozone, 2))
df[["PT08.S5(O3)", "High_Ozone"]].head()
```

## Задание 20

## Задание 21

```python
mean_co = df["CO(GT)"].mean()
above_mean_co = df[df["CO(GT)"] > mean_co]
print("Среднее значение CO(GT):", round(mean_co, 2))
print("Количество измерений выше среднего:", len(above_mean_co))
above_mean_co
```

## Задание 22

```python
mean_temperature = df["T"].mean()
benzene_above = df.loc[df["T"] > mean_temperature, "C6H6(GT)"].mean()
benzene_below = df.loc[df["T"] < mean_temperature, "C6H6(GT)"].mean()

print("Средняя температура:", round(mean_temperature, 2))
print("Средний C6H6(GT) при температуре выше средней:", round(benzene_above, 2))
print("Средний C6H6(GT) при температуре ниже средней:", round(benzene_below, 2))
```
